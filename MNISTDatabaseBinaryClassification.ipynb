```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', as_frame = False) #is an sk.util.bunch object, which has a description, data, and target/labels
X, y = mnist.data, mnist.target
#print(X)
#print(X.shape) # is 784 bc each is 28 by 28 pixels, 70k images
#print(y)
import matplotlib.pyplot as plt
def plot_digit(image_data):
  image = image_data.reshape(28, 28)
  plt.imshow(image, cmap = "binary")#digit is/isn't there
  plt.axis = ("off")#we just want to see the digit
some_digit = X[38] # check by using y(38)
plot_digit(some_digit)
#plt.show()

#MNIST is split into 60k for training, 10k for testing

X_train, X_test, y_train, y_test = X[:60000],X[60000:],y[:60000],y[60000:] #:# is before #: is after
#training dataset is already shuffled which is good for us

#I'm going to use a binary classifier for this

y_train_3 = (y_train == '3')
y_test_3 = (y_test == '3')

#target vector is 3

from sklearn.linear_model import SGDClassifier #imports the stochastic gradient classifier, which is used in linear regression(uses calculus to find the best place for function)
sgd_clf = SGDClassifier(random_state=41)
sgd_clf.fit(X_train, y_train_3)

sgd_clf.predict([some_digit])
#7 is not equal to 3, so should return false
```
