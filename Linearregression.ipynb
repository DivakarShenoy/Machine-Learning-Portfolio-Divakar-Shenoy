```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
reg = LinearRegression()

X = [[1],[2],[3],[4],[5],[6]] #we use 2d arrays b/c each array is an easy way of seperating the entire data point
Y = [10, 20, 30, 41, 53, 59]
reg.fit(X,Y)
print(reg.predict([[5.5],[6.1]])) #input has to be a 2d array b/c it could be multiple

df = pd.DataFrame({'Input': X, 'Output': Y})

fig, ax = plt.subplots(figsize=(8,6))
ax.scatter(X, Y)
num = len(df)
start = np.array(X).min()
end = np.array(X).max()
xseq = np.linspace(start, end, num = num)
plt.show()
plt.title("Y vs X predicted")
plt.xlabel = "X"
plt.ylabel = "Y"
print(reg.score(X,Y))
print(reg.coef_)
intercept = reg.intercept_
ax.plot(xseq, reg.intercept_*reg.coef_[0]*xseq, color = "black")
```

```python
#Dataset: https://www.kaggle.com/datasets/krishnaraj30/salary-prediction-data-simple-linear-regression
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

salary_df = pd.read_csv('/content/sample_data/Salary Data.csv')
salary_df.describe();
X = salary_df['YearsExperience']
Y = salary_df['Salary']

plt.scatter(X,Y)
plt.xlabel("Years Experience")
plt.ylabel("Wage")
plt.title("Graph of Wage vs Years Experience")


learning_rate = 0.01
iterations = 10000
#definite linear trend

def cost(x,y,w,b): #this is the cost function
  m = len(x) #the number of times the MSE should iterate, basically the number of x values
  error = 0 #this is the cost_sum,
  for i in range(m):#iterates through every x-value
     f = w*x[i] + b #w is slope b, is intercept, then this is the predicted value or y hat
     error = (f - y[i]) ** 2 #takes the mean squared error at that poitn
  return (1/(2*m))*error
def gradient(x,y,w,b): #this only make sense after taking partial derivative of the cost function
  m = len(x)
  dc_dw = 0
  dc_db = 0
  for i in range(m):
    f = w*x[i] + b
    dc_dw += (f - y[i]) * x[i] #derivative of ax is a so multiply by it
    dc_db += (f - y[i])
  dc_dw = (1/m) * dc_dw
  dc_db = (1/m) * dc_db
  return dc_dw, dc_db
def gradient_descent(x,y,alpha,iterations):
  w = 0
  b = 0
  for i in range(iterations):
    dc_dw, dc_db = gradient(x,y,w,b)
    w = w - alpha * dc_dw
    b = b - alpha * dc_db
    print("Iteration " + str(w) + " " + str(b))
  return w,b

final_w, final_b = gradient_descent(X, Y, learning_rate, iterations)
```
